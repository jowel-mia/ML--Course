```python
sns.countplot(x="Gender", data=df)
plt.title("Gender Distribution")
plt.show()
```

```python
plt.hist(df["Age"], bins=10)
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")
plt.show()
```

```python
sns.boxplot(x="Education", y="Monthly_Income", data=df)
plt.title("Education vs Monthly Income")
plt.show()
```

```python
df.groupby("Occupation")["Monthly_Income"].mean()
```

```python
sns.barplot(x="Occupation", y="Monthly_Income", data=df)
plt.xticks(rotation=45)
plt.title("Average Income by Occupation")
plt.show()
```

```python
sns.scatterplot(x="Age", y="Monthly_Income", data=df)
plt.title("Age vs Monthly Income")
plt.show()
```

```python
sns.countplot(x="Age_Group", data=df)
plt.title("Age Group Distribution")
plt.show()
```
